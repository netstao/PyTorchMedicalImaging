# 负对数似然损失
1. 分类模型中，负对数似然损失（Negative Log-Likelihood Loss）是衡量模型预测概率分布与真实数据标签之间差异的一种方式。
2. NLLLoss是PyTorch中用于计算负对数似然损失的函数，适用于分类问题。
3. 是交叉熵损失（Cross Entropy Loss）的一种变体，主要用于处理概率分布的预测。加了softmax损失函数，在分类问题中，通常会将模型的输出通过softmax函数转换为概率分布。

```python
import torch 
import torch.nn as nn
import torch.optim as optim

model = nn.Sequential(
    nn.Linear(3072,512),
    nn.Tanh(),
    nn.Linear(512, n_out),
    nn.LogSoftmax(dim=1)
)
learning_rate = 1e-2
optimizer = optim.SGD(model.parameters(), lr=learning_rate)
loss_fn = nn.NLLLoss()
n_epoches = 10
```

```python
for epoch in range(n_epoches):
    for img, label in cifar2:
        out = model(img.view(-1).unsqueeze(0))
        loss = loss_fn(out, torch.tensor([label]))
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    print(f'Epoch {epoch+1}, Loss: {float(loss)}')
```

```python
train_loader = torch.utils.data.DataLoader(cifar2, batch_size=64, shuffle=True)
model = nn.Sequential(
    nn.Linear(3072,512),
    nn.Tanh(),
    nn.Linear(512, n_out),
    nn.LogSoftmax(dim=1)
)
learning_rate = 1e-2
optimizer = optim.SGD(model.parameters(), lr=learning_rate)
loss_fn = nn.NLLLoss()
n_epoches = 150
```

```python
for epoch in range(n_epoches):
    for imgs, labels in train_loader:
        batch_size = imgs.shape[0]
        out = model(imgs.view(batch_size, -1))
        loss = loss_fn(out, labels)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
    print(f'Epoch {epoch+1}, Loss: {float(loss)}')
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

train_loader = torch.utils.data.DataLoader(cifar2, batch_size=64, shuffle=True)

model = nn.Sequential(
            nn.Linear(3072, 1024),
            nn.Tanh(),
            nn.Linear(1024, 512),
            nn.Tanh(),
            nn.Linear(512, 128),
            nn.Tanh(),
            nn.Linear(128, 2))

learning_rate = 1e-2

optimizer = optim.SGD(model.parameters(), lr=learning_rate)

loss_fn = nn.CrossEntropyLoss()

n_epochs = 100

for epoch in range(n_epochs):
    for imgs, labels in train_loader:
        outputs = model(imgs.view(imgs.shape[0], -1))
        loss = loss_fn(outputs, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print("Epoch: %d, Loss: %f" % (epoch, float(loss)))
```

```python
val_loader = torch.utils.data.DataLoader(cifar2_val, batch_size=64, shuffle=False)

correct = 0
total = 0
with torch.no_grad():
    for imgs, labels in val_loader:
        batch_size = imgs.shape[0]
        outputs = model(imgs.view(batch_size, -1))
        _ , predicted = torch.max(outputs, 1)
        total+= labels.shape[0]
        correct +=int((predicted == labels).sum())

print("Accuracy: ", correct /total)
```

# 卷积网络
1. 全连接网络的局限
    * 第一个问题可能是参数太多，导致训练时间长，容易过拟合
    * 第二个问题就是不具有平移不变性，移动后特征就不匹配了
1. 卷积网络
    * 把图像一个区域与周围上下左右的区域关联的这种特性学习的，也就是实现平移不变性,通俗来理解就是，不管图像如何移动，只要图像内容不变，卷积网络就能够识别出图像的内容,一个好一点的办法就是在一个点上，把它周期的点都加起来放在这个点上，当作这个点的数据
    * 周边元素的局部操作
    * 平移不变性
    * 使用很少量的模型参数
