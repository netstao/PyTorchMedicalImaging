# AUC-ROC 曲线
* ROC曲线就是假设我们对判断阳性取不同的阈值时，以两个值为横纵坐标得到的一条曲线。下图中，我们假设使用一个结节的直径大小作为判断是否恶性结节的标准，那么该曲线就是当取不同的直径大小作为判断阈值时，所得到的ROC曲线。

# AUC 值
* AUC 即为ROC曲线下的面积。ROC曲线时用来衡量分类器的分类能力，AUC表示，随机抽取一个正样本和一个负样本，分类器正确给出正样本的score高于负样本的概率
    1. AUC=1 是完美分类器，不存在完美分类器
    2. 0.5>AUC<1 优于随机猜测。这个分类器设定阈值的话，能有预测价值
    3. AUC = 0.5 跟随机猜测一样，模型基本没有预测价值
    4. AUC < 0.5 比随机猜测还差，但总是反预测就行，就优于随机猜测
