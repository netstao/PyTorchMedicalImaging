# DICOM 图像
1. 医学领域中，DICOM（Digital Imaging and Communications in Medicine）是一种广泛使用的标准格式，用于存储、传输和打印医学图像。
2. 就像http之于互联网世界的重要程度一样，DICOM也是一种标准化的通信协议，用于在医疗环境中传输图像数据。

```python
import os
import imageio
import torch
dir_path = '../data/3/dicom'
vol_arr = imageio.volread(dir_path,'DICOM')
vol_arr.shape
```

```python
vol = torch.from_numpy(vol_arr).float()
vol = torch.unsqueeze(vol, 0)
vol.shape
```
