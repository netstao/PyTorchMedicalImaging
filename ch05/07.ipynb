```python
import numpy as np
import torch

torch.set_printoptions(edgeitems=2,threshold=50)

with open('../data/4/1342-0.txt', encoding='utf-8') as f:
    text = f.read()
    lines = text.split('\n')
    line = lines[200]

line
```

```python
letter_t = torch.zeros(len(line),128)
letter_t.shape
```

```python
for i, letter in enumerate(line.lower().strip()):
    letter_index = ord(letter) if ord(letter) < 128 else 0
    letter_t[i][letter_index] = 1
```

# one-hot 编码问题
1. 在文本领域，one-hot编码过于稀疏，在文本场景计算开销过大
2. one-hot编码中，词与词之间是独立的，无法体现词与词之间的相关性

# 词嵌入技术
1. 我们希望找到一个限定维度的空间表示，即可以压缩编码范围，又可以提现词与词之间的关系。

# one-hot 编码的变种
1. 独热编码（One-Hot Encoding）
2. 嵌入编码（Embedding Encoding）
3. 计数编码（Count Encoding）
4. 目标编码（Target Encoding）
5. 均值编码（Mean Encoding）
6. 频率编码（Frequency Encoding）
7. 逆向编码（Backward Difference Encoding）
8. 逆向填充编码（Backward Fill Encoding）
9. 逆向填充均值编码（Backward Fill Mean Encoding）
10. 逆向填充频率编码（Backward Fill Frequency Encoding）
11. 逆向填充最大值编码（Backward Fill Max Encoding）
12. 逆向填充最小值编码（Backward Fill Min Encoding）
13. 逆向填充中位数编码（Backward Fill Median Encoding）
