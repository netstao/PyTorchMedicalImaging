# 激活函数
1. 不使用激活函数，无论神经网络有多少层，输出都是输入值的线性函数。线性组合
2. 激活函数是非线性的，这是非常重要特性之一，非线性使得整个神经网络可以逼近任何复杂函数模型。
3. 激活函数是可微的
* 常见激活函数
1. Sigmoid函数
2. Tanh函数
3. Relu函数
4. Leaky Relu函数
5. Softmax函数
6. ELU函数
7. Softplus函数
8. Swish函数



## 神经元细胞
1. 神经元突起有树突和轴突两种，树突短而分枝多，直接由细体扩张突出，形成树枝状，其作用是接受其他神经元轴突传来的冲动并传给细胞体。轴突长而分枝少，为粗细均匀的细长突起，常起于轴丘，其作用是接受外来刺激，再由细胞体传出。
