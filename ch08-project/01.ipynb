# 肺癌检测项目
1. 辅助医生, 降低医生工作量, 提高诊断效率

## 确定项目背景与目标
1. 拿到一些肺部CT图像数据集
2. 按医生的思路处理这些数据
3. 找到一些异常点位
4. 确认点位是否真的存在问题
5. 最后对存在问题点位进行深入研究确定是否恶性
6. 确定若干疑似情况并进行标注，并交由医生确认 

## DICOM 数据
* CT扫描的每一个体素都对应一个数值，这个数值描述的是内部物质的平均质量密度
## 体素数据与CT值
* 对于一块体素数据，它不是一个立方体，而是1.125mmx1.125mmx2.5mm的一个立体块，如果按照立方体来对数据进行绘制的时候，人会看起来更胖一些，所以如果要按真实的样貌进行展示还需要加入一个变换比例
* CT值是测定人体某一局部组织或器官密度大小的计量单位，称作亨氏单位（Hounsfield unit，简称HU），空气的CT值为-1000，水的CT值为0，骨骼的CT值约为+1000。实际CT值是CT图像中各组织与X衰减系数相当的对应值。

## 观察数据
1. CT数据是立体3D数据
2. 数据存在重复和错误
3. 数据的标注包括一份某个区域的结节是否为肿瘤以及某个肿瘤是否恶性的标准

## 理解业务
1. 拿到CT先逐行逐个查看图像
2. 找到其中存在异常的位置，然后用笔把这个位置圈出来留待分析
3. 对所有圈出来的位置进行仔细的研究，根据它的各种特性，比如光泽，形状，大小，位置等信息来决策是否真的是一个肿瘤
4. 做更多的检查，结合经验判断这个肿瘤是否是恶性的肿瘤
5. 帮助病人制定治疗 方案

## 解决方案
1. 拿到数据后，先对数据进行预处理，去掉重复的和错误的 数据转换
2. 全像素空间中寻找有问题的像素
3. 对有问题的点位进行分组，对结节的图像进行分类，确定是否为肿瘤
4. 对肿瘤进行分类，确定是否为恶性
#### PyTorch
1. 加载原始数据，把它转为为PyTorch的Tensor格式
2. 使用PyTorch进行数据分割，把存在异常的点位确定出来
3. 第三步是对结节进行分组
4. 构建模型对候选结节进行分类，确定是否为肿瘤
5. 对分类结果进行分析，然后根据影像判断肿瘤是良性还是恶性
