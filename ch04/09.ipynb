# 转置

```python
points = torch.tensor([[4,1],[5,3],[2,1]])
points

point_t = points.t()
point_t
```

```python
points = torch.tensor([[4,1],[5,3],[2,1]])

point_t = points.t()
point_t
id(point_t.storage()) == id(points.storage())
```

```python
points.stride()
```

```python
point_t.stride()
```

```python
points.is_contiguous()
```
