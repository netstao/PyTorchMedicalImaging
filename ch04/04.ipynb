# 张量的基本操作
1.张量的拼接

```python
a = torch.ones((4,5))
a
```

```python
cat_a = torch.cat([a,a], dim=0)
cat_a
```

```python
cat_a2 = torch.cat([a,a], dim=1)
cat_a2
```

```python
a = torch.normal(0,1,(3,4))
b = torch.normal(0,1,(3,4))
```

```python
cat_ab1 = torch.stack([a,b], dim=0)
cat_ab1
```

```python
cat_ab2 = torch.stack([a,b], dim=2)
cat_ab2
```

# 张量的切分

```python
a = torch.ones((3,7))
chunked_a = torch.chunk(a,3,1)
chunked_a
```

```python
split_a1 = torch.split(a,4,1)
split_a1
```

```python
points = torch.tensor([[4,1],[5,4],[2,1]])
idx = torch.tensor([0,2])
res = torch.index_select(points, dim=0,index=idx)
res
```

```python
mask = points.ge(3)
mask
```

```python
res = torch.masked_select(points,mask)
res
```
